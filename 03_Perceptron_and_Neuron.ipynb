
## 1. Biological Neuron vs Artificial Neuron

**Explain:** A **biological neuron** receives electrical signals through dendrites, combines them in the cell body, and fires an output signal through its axon if the combined signal is strong enough. An **artificial neuron** mimics this loosely: it takes numeric inputs, combines them mathematically, and produces a numeric output if the combined value is "strong enough" (via an activation function).

**Analogy:** A biological neuron is like a single voter listening to several arguments (signals) — if the combined persuasion is strong enough, they "fire" (vote yes). An artificial neuron does the same thing with numbers instead of arguments.

**Implement:** a simple side-by-side comparison table.

```python
comparison = {
    "Signal": ("Electrical/chemical", "Numeric input"),
    "Combining": ("Dendrites sum signals", "Weighted sum (w1x1+w2x2+...+b)"),
    "Firing rule": ("Fires if threshold exceeded", "Activation function (e.g. sigmoid, ReLU)"),
    "Output": ("Electrical pulse via axon", "A number passed to the next neuron"),
}
for k, (bio, art) in comparison.items():
    print(f"{k:10} | Biological: {bio:30} | Artificial: {art}")
```

**Interpret:** The artificial neuron is a deliberately simplified, mathematical version of the biological one — same basic idea (combine inputs, decide whether to "fire"), but expressed with numbers and functions instead of biochemistry.

## 2. Perceptron

**Explain:** A **perceptron** is the simplest possible artificial neuron — it takes inputs, computes a weighted sum plus bias, and applies a **step activation** (output 1 if the sum is positive, else 0). It was one of the earliest neural network models (1958) and is the building block every modern network is based on.

**Analogy:** A perceptron is like a bouncer at a club with one simple yes/no rule based on a few weighted factors (age, dress code, guest list) — no nuance, just a hard yes or no.

**Implement:**

```python
def perceptron(x, w, b):
    z = np.dot(x, w) + b
    return 1 if z > 0 else 0     
print(perceptron(x=[1, 1], w=[1, 1], b=-0.5))
```

**Interpret:** With these weights and bias, the perceptron outputs `1` — a simple yes/no decision based on the weighted combination of inputs crossing zero.


## 3. Inputs

**Explain:** **Inputs** (`x1, x2, ...`) are the raw numeric signals fed into the perceptron — the "facts" it bases its decision on.

**Analogy:** The bouncer's checklist items: is the guest on the list? Are they over 18?

**Implement:**

```python
x1, x2 = 1, 0
print("Inputs:", x1, x2)
```

**Interpret:** These two numbers are everything the perceptron "knows" — its decision can only depend on these values, combined with weights and bias.

## 4. Weights

**Explain:** **Weights** (`w1, w2, ...`) scale how much each input matters. A larger weight means that input has more influence on the final decision.

**Analogy:** How strictly the bouncer cares about each rule — being on the guest list might matter much more (higher weight) than dress code.

**Implement:**

```python
w1, w2 = 2.0, 0.5
print("x1 contributes:", x1 * w1, "| x2 contributes:", x2 * w2)
```

**Interpret:** Even though `x1` and `x2` are just 1 and 0, the weight makes `x1`'s contribution much larger — this is exactly how weights control each input's importance.


## 5. Bias

**Explain:** The **bias** (`b`) is a constant added to the weighted sum, shifting the decision threshold. It lets the perceptron fire (or not) even when inputs alone wouldn't cross zero.

**Analogy:** A bouncer who's naturally lenient (positive bias) or strict (negative bias), independent of the specific guest in front of them.

**Implement:**

```python
b = -1.0
z_with_bias = (x1 * w1 + x2 * w2) + b
print("Weighted sum + bias:", z_with_bias)
```

**Interpret:** The bias shifted the total down by 1 — without it, the decision boundary would sit exactly at zero; with it, the boundary moves, letting us fine-tune when the perceptron fires.

## 6. Weighted Sum

**Explain:** The **weighted sum** (`z = w1x1 + w2x2 + ... + b`) combines every input, weight, and the bias into a single number — the perceptron's "raw score" before any activation is applied.

**Analogy:** A bouncer's mental tally after checking every rule, before making the final yes/no call.

**Implement:**

```python
z = w1 * x1 + w2 * x2 + b
print("z =", z)
```

**Interpret:** This single number `z` captures everything the perceptron has "considered" so far — the next step decides what to actually do with it.


## 7. Activation Function

**Explain:** The **activation function** turns the raw weighted sum `z` into the final output — e.g. a step function (1 if z>0, else 0), or smoother functions like sigmoid. It's what turns a plain linear calculation into a decision.

**Analogy:** The bouncer's final "yes, come in" or "no, stay out" — turning their mental tally into one concrete action.

**Implement:**

```python
def step(z):
    return 1 if z > 0 else 0
print("Activation output:", step(z))
```

**Interpret:** The raw score `z` (1.0, from Section 6) was positive, so the step function converted it into a `1` — the perceptron "decided" to fire, since the combined weighted evidence crossed the zero threshold.


## 8. Output

**Explain:** The **output** is the final result the perceptron produces after activation — this is what gets used as the prediction, or passed on to the next neuron in a bigger network.

**Analogy:** The bouncer physically letting someone in, or turning them away — the final, visible outcome.

**Implement:**

```python
final_output = step(w1 * x1 + w2 * x2 + b)
print("Final perceptron output:", final_output)
```

**Interpret:** This `0` or `1` is the end of the line for this single perceptron — in a full network, this number would become one of the *inputs* to the next layer.

## 9. Mathematical Representation

**Explain:** Putting it all together, a perceptron computes:

`z = w1*x1 + w2*x2 + ... + b`

`output = activation(z)`

This is the exact same formula for **every** neuron in **every** neural network — only the number of inputs, the specific weights/bias, and the activation function change.

**Implement:** the same formula, written generally for any number of inputs using NumPy's dot product.

```python
def neuron(x, w, b, activation=step):
    z = np.dot(x, w) + b
    return activation(z)
print(neuron(x=[1, 1], w=[1, 1], b=-0.5))
```

**Interpret:** `np.dot(x, w)` computes `w1*x1 + w2*x2 + ...` in one line, no matter how many inputs there are — this generality is why the same formula scales from 2 inputs to millions.


## 10. Implement a Simple Neuron Manually (NumPy)

**Explain:** Let's build one complete neuron from scratch — no shortcuts — to see every piece (inputs, weights, bias, weighted sum, activation, output) working together on a realistic-looking input.

**Implement:**

```python
inputs = np.array([0.6, 0.9, 0.2])    
weights = np.array([0.4, -0.7, 0.3])
bias = 0.1
z = np.dot(inputs, weights) + bias
output = step(z)
print(f"z = {z:.2f}  ->  output = {output}")
```

**Interpret:** This is a complete, working artificial neuron — every concept from Sections 3–8 (inputs, weights, bias, weighted sum, activation, output) is present in these 3 lines of math.


## 11. AND Gate

**Explain:** An **AND gate** outputs `1` only when **both** inputs are `1`. It's a classic first test for a perceptron because it's **linearly separable** — one straight line can separate the "1" outputs from the "0" outputs.

**Analogy:** A door that only opens if **both** keyholders turn their key at the same time.

**Implement:** hand-pick weights and bias that make a perceptron behave like an AND gate.

```python
and_inputs = [(0,0), (0,1), (1,0), (1,1)]
w_and, b_and = [1, 1], -1.5     
for x in and_inputs:
    print(x, "->", neuron(x, w_and, b_and))
```

**Interpret:** The perceptron outputs `1` only for `(1,1)` — exactly matching the AND truth table. A bias of `-1.5` means the sum of two inputs (max 2) must be both 1s to push `z` above 0.


## 12. OR Gate

**Explain:** An **OR gate** outputs `1` if **at least one** input is `1`. It's also linearly separable — a different straight line (different bias) separates its outputs.

**Analogy:** A door that opens if **either** keyholder turns their key.

**Implement:** same neuron, different bias.

```python
w_or, b_or = [1, 1], -0.5     
for x in and_inputs:
    print(x, "->", neuron(x, w_or, b_or))
```

**Interpret:** Now the perceptron outputs `1` for any input pair except `(0,0)` — exactly the OR truth table. Only the bias changed, yet the perceptron now represents a completely different logical rule.


## Why a Single Perceptron Cannot Solve Every Problem (the XOR Problem)

**Explain:** AND and OR both work because their "1" outputs and "0" outputs can be separated by **one straight line**. A single perceptron can only draw straight-line (linear) decision boundaries — no matter how we set its weights and bias. **XOR** (output `1` only when the inputs *differ*) cannot be separated by any single straight line — its "1"s and "0"s are arranged diagonally, so no single perceptron can ever learn it correctly.

**Analogy:** Imagine 4 dots arranged in a square, with the two diagonal corners colored red and the other two colored blue. No single straight line can separate red from blue — you'd need at least two lines (i.e. more than one neuron / a hidden layer) to do it.

**Implement:** try every possible straight-line perceptron (loop over many weight/bias combinations) and show that none can solve XOR — while confirming AND and OR work.

```python
xor_targets = [0, 1, 1, 0]     
and_targets = [0, 0, 0, 1]
best_xor_correct = 0
for w1_ in np.arange(-2, 2.1, 0.5):
    for w2_ in np.arange(-2, 2.1, 0.5):
        for b_ in np.arange(-2, 2.1, 0.5):
            preds = [neuron(x, [w1_, w2_], b_) for x in and_inputs]
            correct = sum(p == t for p, t in zip(preds, xor_targets))
            best_xor_correct = max(best_xor_correct, correct)

print("Best a single perceptron can do on XOR (out of 4):", best_xor_correct)
print("(AND is solved perfectly with just 1 line, as shown above)")
```

**Interpret:** Even after searching many weight/bias combinations, a single perceptron gets **at most 3 out of 4** XOR cases right — it can never solve all 4, because XOR is **not linearly separable**. This is the famous limitation that (temporarily) stalled neural network research in the 1960s-70s, and it's exactly why modern networks use **hidden layers**: stacking multiple perceptrons together lets the network draw multiple lines (a curved/combined boundary), which *can* solve XOR — this is the topic of the next notebook.
