## 1. Deep learning
Deep learning uses neural networks with layers to learn patterns from examples. Training adjusts weights to reduce mistakes.

## 2. Neural network
A neural network connects input values to hidden and output layers. Each layer learns a useful pattern.

## 3. Perceptron, weights, and bias
A perceptron calculates a weighted sum plus a bias. With `x = [2, -1]`, `w = [0.5, 1]`, and `b = 0.25`, the sum is `2(0.5) - 1 + 0.25 = 0.25`, so a step activation returns 1.

```python
x = [2, -1]
weights = [0.5, 1]
bias = 0.25
score = sum(a * b for a, b in zip(x, weights)) + bias
print(score, int(score >= 0))
```

## 4. Activation functions
ReLU returns zero for negative values and keeps positive values. Sigmoid returns a value from 0 to 1 and is often used for a binary probability.

```python
import torch
print(torch.relu(torch.tensor([-1.0, 2.0])))
print(torch.sigmoid(torch.tensor([0.0])))
```

## 5. Forward propagation
Forward propagation sends inputs through each layer to produce an answer. For `h = ReLU(3 × 2 + 1)`, `h = 7`; if output is `2h - 1`, the answer is 13.

```python
hidden = torch.relu(torch.tensor(3.0 * 2 + 1))
print(2 * hidden - 1)
```

## 6. Loss functions
Loss measures how far predictions are from correct answers. Mean squared error is common for numeric predictions; cross-entropy is common for classification. For targets `[1, 0]` and predictions `[0.8, 0.4]`, MSE is 0.1.

```python
targets = torch.tensor([1.0, 0.0])
predictions = torch.tensor([0.8, 0.4])
print(torch.mean((targets - predictions) ** 2).item())
```

## 7. Gradient descent and backpropagation
Gradient descent changes weights in the direction that lowers loss. Backpropagation calculates the gradients using the chain rule. If `w = 2`, gradient is 3, and learning rate is 0.1, the updated weight is 1.7.

```python
weight = 2.0
weight = weight - 0.1 * 3.0
print(weight)
```

## 8. Optimizers
SGD updates weights using gradients. Adam adapts update sizes using recent gradients. Try a smaller learning rate if training is unstable.

```python
import torch.nn as nn
model = nn.Linear(2, 1)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
print(type(optimizer).__name__)
```

## 9. PyTorch tensors
A tensor stores numbers and has a shape. The example has two rows and two columns; its sum is 10.

```python
values = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
print(values.shape, values.sum().item())
```

## 10. Autograd
Autograd calculates derivatives for tensors that track gradients. For `y = x²` at `x = 3`, the derivative is 6.

```python
x = torch.tensor(3.0, requires_grad=True)
(x ** 2).backward()
print(x.grad.item())
```

## 11. Dataset and DataLoader
A Dataset provides one example and its label. A DataLoader makes batches and can shuffle examples during training.

```python
from torch.utils.data import TensorDataset, DataLoader
dataset = TensorDataset(torch.tensor([[1.], [2.]]), torch.tensor([0, 1]))
batch = next(iter(DataLoader(dataset, batch_size=2)))
print(batch)
```

## 12. ANN and CNN
An ANN is useful for tabular values. A CNN is useful for images because it learns nearby pixel patterns.

The ANN below maps four inputs to two classes. The CNN accepts a one-channel image.

```python
ann = nn.Sequential(nn.Linear(4, 8), nn.ReLU(), nn.Linear(8, 2))
cnn = nn.Conv2d(1, 4, kernel_size=3, padding=1)
print(ann)
print(cnn)
```

## 13. Overfitting and regularization
Overfitting means training results are much better than validation results. Dropout and weight decay are two ways to help the model generalize.

```python
regularized_model = nn.Sequential(nn.Linear(4, 8), nn.ReLU(), nn.Dropout(0.2), nn.Linear(8, 1))
optimizer = torch.optim.Adam(regularized_model.parameters(), weight_decay=0.001)
print(regularized_model)
```

## 14. Transfer learning
Transfer learning starts with a model already trained on another dataset. Freeze its learned layers and train a new output layer. This is useful when the new dataset is small.

```python
feature_layer = nn.Linear(4, 8)
feature_layer.requires_grad_(False)
classifier = nn.Linear(8, 2)
transfer_model = nn.Sequential(feature_layer, nn.ReLU(), classifier)
print(transfer_model)
```

## 15. Model evaluation
Accuracy is the share of correct predictions. Precision measures how many predicted positives were correct. Recall measures how many real positives were found. Use recall when missing positives is costly.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score
actual = [1, 1, 0, 0]
predicted = [1, 0, 1, 0]
print('Accuracy:', accuracy_score(actual, predicted))
print('Precision:', precision_score(actual, predicted))
print('Recall:', recall_score(actual, predicted))
```

## 16. Training-step answer
Clear old gradients, calculate loss, backpropagate, then update the weights.

```python
model = nn.Linear(2, 1)
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
inputs = torch.tensor([[1.0, 2.0]])
target = torch.tensor([[1.0]])
optimizer.zero_grad()
loss = (model(inputs) - target).square().mean()
loss.backward()
optimizer.step()
```
